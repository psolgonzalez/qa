Imprimir por pantalla los números del 1 al 100 pero considerando lo siguiente: a) Si el número es divisible por 3 se debe imprimir "Fizz". b) Si el número es divisible por 5 se debe imprimir "Buzz". c) Si el número es divisible por 3 y por 5 se debe imprimir "FizzBuzz"

```python
for numero in range(1, 101):
    if numero % 3 == 0 and numero % 5 == 0:
        print("FizzBuzz")   # ✅ con doble z
    elif numero % 3 == 0:
        print("Fizz")
    elif numero % 5 == 0:
        print("Buzz")
    else:
        print(numero)
       

```

```python
def es_divisible(numero, divisor):
    return numero % divisor == 0

def fizzbuzz(numero):
    if es_divisible(numero, 3) and es_divisible(numero, 5):
        return 'FizzBuzz'
    if es_divisible(numero, 3):
        return 'Fizz'
    if es_divisible(numero, 5):
        return 'Buzz'
    return numero

def get_lista(numero):
    lista = []
    for i in range(1, numero+1):
        lista.append(fizzbuzz(i))
    return lista
```

Los números expansivos se definen de la siguiente forma:
* El primer número expansivo es el 1
* Dado un número expansivo x, para calcular el siguiente se ha de sustituir cada grupo de n cifras consecutivas iguales que aparezcan en x (por ejemplo, 3333) por número n seguido de la cifra que se repite (en el ejemplo 43)
* Así, los primeros números expansivos son:
* 1 -> "un uno": 11
* 11 -> "dos unos": 21
* 21 -> "un dos, un uno": 1211
* 1211 -> "un uno, un dos, dos uno": 111221

Se pide programar: 
1. una función "expand" que, dado un arreglo con las cifras de un número expansivo genere el siguiente
2. una función "list2num" que, dado un arreglo con las cifras de un número expansivo, devuelva el número correspondiente

```python
def expand(lista):
    resultado = []
    i = 0
    while i < len (lista):
        digito = lista[i]
        cantidad = 1
        # cuenta las veces q repite el mismo numero 
        while i + 1 < len(lista) and lista[i+1] == digito:
            cantidad += 1
            i += 1
        resultado.append(cantidad)
        resultado.append(digito)
        i += 1
    return resultado

print(expand([1]))       

print(expand([1,1]))     
print(expand([2,1])) 

print(expand([1,2,1,1])) 
```

```python
def list2num(lista):
    numero = 0
    for digito in lista:
        numero = numero * 10 + digito
    return numero

print(list2num([1,1]))        
print(list2num([2,1]))       
print(list2num([1,2,1,1]))    

```

```python
class CuentaBancaria:

    def __init__(self, saldo_inicial=0):
        self.__saldo = saldo_inicial

    def get_saldo(self):
        return self.__saldo

    def depositar(self, monto):
        self.__saldo += monto

    def retirar(self, monto):
        if monto > self.__saldo:
            raise ValueError("Fondos insuficientes")
        self.__saldo -= monto

cuenta = CuentaBancaria(100)   # saldo 100
print(cuenta.get_saldo())      

cuenta.depositar(50)
print(cuenta.get_saldo())     

cuenta.retirar(30)
print(cuenta.get_saldo())      # 120

# Si intentás retirar más de lo que hay:
cuenta.retirar(200)            # ValueError: Fondos insuficientes

```

```python
import unittest

class CuentaBancariaTest(unittest.TestCase):
    def test_saldo_inicial(self):
        cuenta = CuentaBancaria(100)
        self.assertEqual(cuenta.get_saldo(), 100)

    def test_depositar(self):
        cuenta = CuentaBancaria(50)
        cuenta.depositar(30)
        self.assertEqual(cuenta.get_saldo(), 80)  # 50 + 30 = 80

    def test_retirar_valido(self):
        cuenta = CuentaBancaria(100)
        cuenta.retirar(40)
        self.assertEqual(cuenta.get_saldo(), 60)

    def test_retirar_insuficiente(self):
        cuenta = CuentaBancaria(50)
        with self.assertRaises(ValueError):
            cuenta.retirar(100)

unittest.main(argv=[''], verbosity=2, exit=False)
```

```python
def es_edad_valida(edad):
    return 0 <= edad <= 120

import unittest

class TestLimite(unittest.TestCase):

    def test_edad_limite(self):
                                # valores en el limite
        self.assertTrue(es_edad_valida(0))
        self.assertTrue(es_edad_valida(120))
                             # valores fuera dellimite
        self.assertFalse(es_edad_valida(-1))
        self.assertFalse(es_edad_valida(121))

unittest.main(argv=[''], verbosity=2, exit=False)
```

```python
""" pruebas con mocks: se usan objetos simulados para probar código
 que depende de algo externo (como enviar correos) sin necesidad de 
 realmente conectarse a un servidor de email.
 el test confirma que tu código intenta enviar un correo, pero no lo manda de verdad."""

class Correo:
    def enviar(self, mensaje):
        # En la vida real mandaría un email
        pass

class Usuario:
    def __init__(self, correo):
        self.correo = correo

    def notificar(self, mensaje):
        self.correo.enviar(mensaje)


import unittest
from unittest.mock import Mock

class TestMocking(unittest.TestCase):
    def test_notificar(self):
        correo_mock = Mock()              # objeto simulado
        usuario = Usuario(correo_mock)    # inyectamos el mock
        usuario.notificar("Hola Pamela")  # llamamos al método
        # verificamos que se haya usado el mock
        correo_mock.enviar.assert_called_with("Hola Pamela")

unittest.main(argv=[''], verbosity=2, exit=False)
```
