```python
def sumar(num1, num2):
    return num1 + num2

def restar(num1, num2):
    return num1 - num2

import unittest

class CuentasTest(unittest.TestCase):
    def test_suma_20_30(self):
        actual = sumar(20, 30)
        esperado = 50
        self.assertEqual(esperado, actual)

    def test_resta_200_100(self):
        actual = restar(200, 100)
        esperado = 100
        self.assertEqual(esperado, actual)

unittest.main(defaultTest="CuentasTest", argv=[''], verbosity=3,exit=False)
```

```python
import unittest

def sucesor(n):
    return n + 1

class SucesorTest(unittest.TestCase):

    def test_1(self):
        self.assertEqual(3, sucesor(2))

    def test_2(self):
        self.assertEqual(1, sucesor(0))

    def test_varios(self):
        casos = [(0, -1), (1, 0), (2, 1)]
        for esperado, valor in casos:
            with self.subTest(valor=valor):
                self.assertEqual(esperado, sucesor(valor))

unittest.main(defaultTest='SucesorTest', argv=[''], verbosity=2, exit=False)
```

Imprimir por pantalla los números del 1 al 100 pero considerando lo siguiente: a) Si el número es divisible por 3 se debe imprimir "Fizz". b) Si el número es divisible por 5 se debe imprimir "Buzz". c) Si el número es divisible por 3 y por 5 se debe imprimir "FizzBuzz"

```python
def fizzbuzz(numero):
    if numero % 3 == 0 and numero % 5 == 0:
        return "FizzBuzz"
    elif numero % 3 == 0:
        return "Fizz"
    elif numero % 5 == 0:
        return "Buzz"
    else:
        return numero

import unittest

class FizzBuzzTest(unittest.TestCase):
    def test_fizz(self):
        self.assertEqual(fizzbuzz(3), "Fizz")

    def test_buzz(self):
        self.assertEqual(fizzbuzz(5), "Buzz")

    def test_fizzbuzz(self):
        self.assertEqual(fizzbuzz(15), "FizzBuzz")

    def test_numero(self):
        self.assertEqual(fizzbuzz(2), 2)

unittest.main(defaultTest="FizzBuzzTest",argv=[''], verbosity=2, exit=False)

```

Los números expansivos se definen de la siguiente forma:
* El primer número expansivo es el 1
* Dado un número expansivo x, para calcular el siguiente se ha de sustituir cada grupo de n cifras consecutivas iguales que aparezcan en x (por ejemplo, 3333) por número n seguido de la cifra que se repite (en el ejemplo 43)
* Así, los primeros números expansivos son:
* 1 -> "un uno": 11
* 11 -> "dos unos": 21
* 21 -> "un dos, un uno": 1211
* 1211 -> "un uno, un dos, dos uno": 111221

Se pide programar: 
1. una función "expand" que, dado un arreglo con las cifras de un número expansivo genere el siguiente
2. una función "list2num" que, dado un arreglo con las cifras de un número expansivo, devuelva el número correspondiente

```python
class Contador:
    def __init__(self):
        self.valor = 0

    def incrementar(self):
        self.valor += 1


import unittest

class TestEstado(unittest.TestCase):
    def test_incrementar(self):
        c = Contador()
        c.incrementar()
        self.assertEqual(1, c.valor)   
        c.incrementar()
        self.assertEqual(2, c.valor)

unittest.main(defaultTest="TestEstado",argv=[''], verbosity=2, exit=False)
```
