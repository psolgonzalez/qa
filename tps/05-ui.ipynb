```python
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium import webdriver

driver = webdriver.Chrome()


driver.get("https://coffee-cart.app/")

#cargo 2 tazas de cafe

driver.find_element(By.XPATH, "//*[@data-test='Cafe_Breve']").click()

#cartelito de 30 pesos
assert driver.find_element(By.XPATH,"//*[@aria-label='Proceed to checkout']").text == 'Total: $15.00'

driver.quit()
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service

driver = webdriver.Chrome()

driver.get("https://coffee-cart.app/")

#agrego un cafe Americano
driver.find_element(By.XPATH, "//*[@data-test='Americano']").click()

#hago click en precio
driver.find_element(By.XPATH,"//*[@class='pay']").click()

  #muestra 'Payment details'
#assert driver.find_element(By.TAG_NAME, "h1").text =='Payment details'
assert driver.find_element(By.XPATH, "//h1").text =='Payment details'

driver.quit()
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
import time

driver = webdriver.Chrome()

driver.get("https://coffee-cart.app/")

#cargo un cafe Cappuccino
driver.find_element(By.XPATH, "//*[@data-test='Cappuccino']").click()

#hago click en precio
driver.find_element(By.XPATH,"//*[@class='pay']").click()

#completo Detalle de pago
driver.find_element(By.ID, 'name').send_keys('pepe')
driver.find_element(By.ID, 'email').send_keys('pepe@gmail.com')
driver.find_element(By.XPATH,"//button[@id='submit-payment']").click()

#espera un segundo
time.sleep(1)

assert driver.find_element(By.XPATH, "//div[@class='snackbar success']").text == 'Thanks for your purchase. Please check your email for payment.'

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By

driver = webdriver.Chrome()
driver.get("https://www.google.com")

assert driver.title == "Google"

# Busco el cuadro de búsqueda con XPath
search_box = driver.find_element(By.XPATH, "//textarea[@name='q']")

assert search_box is not None

driver.quit()
```
